```python
import torch
scores = torch.randn((16, 84, 80))
```

```python
scores1 = scores.amax(dim=-1)
scores1.shape
```

```python
scores2 = scores1.topk(10)[1]
scores2.shape
```

```python
scores3 = scores2.unsqueeze(-1)
scores3.shape
```

```python
boxes = torch.randn((16, 84, 4))
```

```python
boxes1 = boxes.gather(dim=1, index=scores3.repeat(1, 1, 4))
boxes1.shape
```
